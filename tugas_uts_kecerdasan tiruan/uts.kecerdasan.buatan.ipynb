## Soal:
Dinas Pariwisata Lombok Tengah ingin mengembangkan aplikasi navigasi jalur sepeda wisata yang efisien bagi pengunjung. Rute utama dari Taman Tastura Praya menuju Pantai Tanjung Aan.

## 1. Analisis Graph

Berdasarkan data Tabel 1.1, berikut adalah struktur graph berarah yang merepresentasikan hubungan antar tempat wisata:

```
Taman Tastura Praya  --2.0 KM-->  Masjid Agung Praya
Masjid Agung Praya   --1.5 KM-->  Pasar Renteng
Pasar Renteng        --3.2 KM-->  Bundaran Adi Soetjipto
Bundaran Adi Soetjipto --2.8 KM--> Kantor Bupati Lombok Tengah
Kantor Bupati Lombok Tengah --1.0 KM--> Alun-Alun Tastura
Alun-Alun Tastura    --2.3 KM-->  Pantai Selong Belanak
Pantai Selong Belanak --12.3 KM--> Pantai Mawun
Pantai Mawun         --8.5 KM-->  Sirkuit Mandalika
Sirkuit Mandalika    --7.4 KM-->  Pantai Tanjung Aan
Pantai Tanjung Aan   (Tujuan Akhir)
```

Setiap simpul (node) mewakili satu lokasi wisata, dan setiap sisi (edge) mewakili jarak dalam kilometer (KM) ke tempat berikutnya.

## 2. Implementasi BFS (Breadth-First Search)

````python
from collections import deque

def bfs_jalur_terpendek(graph, start, goal):
    # queue berisi: (node_saat_ini, jalur_yang_dilalui, total_jarak)
    queue = deque([(start, [start], 0.0)])
    visited = set()

    while queue:
        current, path, total_dist = queue.popleft()

        if current in visited:
            continue

        visited.add(current)

        if current == goal:
            return path, total_dist

        neighbors = graph.get(current, {})
        for neighbor, distance in neighbors.items():
            if neighbor not in visited:
                new_path = list(path)
                new_path.append(neighbor)
                queue.append((neighbor, new_path, total_dist + distance))

    return None, 0.0


# Struktur graph berdasarkan Tabel 1.1
graph = {
    'Taman Tastura Praya': {'Masjid Agung Praya': 2.0},
    'Masjid Agung Praya': {'Pasar Renteng': 1.5},
    'Pasar Renteng': {'Bundaran Adi Soetjipto': 3.2},
    'Bundaran Adi Soetjipto': {'Kantor Bupati Lombok Tengah': 2.8},
    'Kantor Bupati Lombok Tengah': {'Alun-Alun Tastura': 1.0},
    'Alun-Alun Tastura': {'Pantai Selong Belanak': 2.3},
    'Pantai Selong Belanak': {'Pantai Mawun': 12.3},
    'Pantai Mawun': {'Sirkuit Mandalika': 8.5},
    'Sirkuit Mandalika': {'Pantai Tanjung Aan': 7.4},
    'Pantai Tanjung Aan': {}
}

# Jalankan BFS
jalur, total_km = bfs_jalur_terpendek(graph, 'Taman Tastura Praya', 'Pantai Tanjung Aan')

# Tampilkan hasil
if jalur:
    print("Jalur yang ditemukan:")
    for i, tempat in enumerate(jalur, 1):
        print(f"{i}. {tempat}")
    print(f"\nTotal Jarak: {total_km}KM")
else:
    print("Jalur tidak ditemukan.")
````

## 3. Analisis Proses Pencarian BFS

### Cara Kerja Algoritma BFS
BFS (Breadth-First Search) adalah algoritma pencarian graf yang bekerja secara **berlapis (level by level)**. Algoritma ini menggunakan struktur data **deque (queue/antrian)** dengan prinsip FIFO (First In, First Out).

### Proses Pencarian pada Kasus Ini

| Langkah | Node Diproses | Tetangga Ditemukan | Total Jarak |
|---------|--------------|-------------------|-------------|
| 1 | Taman Tastura Praya | Masjid Agung Praya | 2.0 KM |
| 2 | Masjid Agung Praya | Pasar Renteng | 3.5 KM |
| 3 | Pasar Renteng | Bundaran Adi Soetjipto | 6.7 KM |
| 4 | Bundaran Adi Soetjipto | Kantor Bupati Lombok Tengah | 9.5 KM |
| 5 | Kantor Bupati Lombok Tengah | Alun-Alun Tastura | 10.5 KM |
| 6 | Alun-Alun Tastura | Pantai Selong Belanak | 12.8 KM |
| 7 | Pantai Selong Belanak | Pantai Mawun | 25.1 KM |
| 8 | Pantai Mawun | Sirkuit Mandalika | 33.6 KM |
| 9 | Sirkuit Mandalika | Pantai Tanjung Aan | 41.0 KM |
| 10 | **Pantai Tanjung Aan** | **GOAL TERCAPAI** | **41.0 KM** |

### Bagaimana Jalur Terpendek Diperoleh
Pada kasus ini, graph bersifat **linear searah** — setiap node hanya memiliki satu tetangga. Oleh karena itu, BFS langsung menemukan satu-satunya jalur yang tersedia tanpa percabangan.

BFS menjamin jalur pertama yang ditemukan adalah jalur dengan **jumlah langkah (hop) paling sedikit**. Karena graph ini tidak bercabang, jalur yang ditemukan sekaligus merupakan jalur terpendek berdasarkan total jarak.

### Kesimpulan
Algoritma BFS berhasil menemukan jalur dari **Taman Tastura Praya** ke **Pantai Tanjung Aan** melewati 10 titik lokasi wisata dengan **total jarak 41.0 KM**. BFS terbukti efektif untuk menelusuri graf berarah seperti rute sepeda wisata Lombok Tengah ini.
